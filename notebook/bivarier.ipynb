```python
import seaborn as sns
import matplotlib.pyplot as plt

features = [c for c in data.select_dtypes(include="number").columns if c != "default"]

corr = data[features].corr()

plt.figure(figsize=(18, 14))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)
plt.title("Matrice de corrélation — Features")
plt.tight_layout()
plt.show()
```
